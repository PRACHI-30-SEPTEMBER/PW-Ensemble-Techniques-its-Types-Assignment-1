# Ensemble Techniques Assignment

Questions and Answers

### Q1. What is an ensemble technique in machine learning?

**Answer:**

Ensemble techniques combine multiple machine learning models to improve overall performance. Instead of relying on a single model, multiple models are used and their outputs are combined to produce a better and more stable prediction.

### Q2. Why are ensemble techniques used in machine learning?

**Answer:**

Ensemble techniques are used to improve accuracy, reduce overfitting, and make models more robust. By combining multiple models, errors of individual models can be minimized.

### Q3. What is bagging?

**Answer:**

Bagging (Bootstrap Aggregating) is an ensemble method where multiple models are trained on different random subsets of data (with replacement), and their predictions are combined to reduce variance.

### Q4. What is boosting?

**Answer:**

Boosting is an ensemble technique where models are trained sequentially, and each new model focuses on correcting the errors made by previous models. Example: AdaBoost, Gradient Boosting.

### Q5. What are the benefits of using ensemble techniques?

**Answer:**

- Improved accuracy
- Reduced overfitting
- Better generalization
- More stable predictions

### Q6. Are ensemble techniques always better than individual models?

**Answer:**

No, ensemble techniques are not always better. They can increase complexity and computation time. In some simple problems, a single model may perform equally well.

### Q7. How is the confidence interval calculated using bootstrap?

**Answer:**

In bootstrap, multiple samples are drawn with replacement from the dataset. For each sample, the statistic (like mean) is calculated. The confidence interval is obtained by taking percentiles (e.g., 2.5% and 97.5%) of these bootstrap estimates.

### Q8. How does bootstrap work and what are the steps involved in bootstrap?

**Answer:**

Steps:
1. Draw random samples with replacement from dataset
2. Train model or calculate statistic
3. Repeat multiple times
4. Aggregate results to estimate distribution
5. Calculate metrics like mean, variance, confidence interval

### Q9. A researcher wants to estimate the mean height...

**Answer:**

Given:
Sample size = 50
Mean = 15
Std deviation = 2

Standard Error = 2/sqrt(50) ≈ 0.283

95% CI = Mean ± 1.96 × SE
= 15 ± 1.96×0.283
= 15 ± 0.554
